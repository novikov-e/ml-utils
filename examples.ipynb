```python
from IPython.display import display, HTML

# ---------------------------------------------------------
# ТЕСТ А: Жёсткие инлайн-стили (СИНЯЯ РАМКА)
# Пытаемся перебить стили Colab, прописывая всё прямо в тегах + !important
# ---------------------------------------------------------
test_a = """
<div style="margin-bottom: 30px;">
    <h3 style="color: black !important;">Тест А (Синий): Инлайн-стили + !important</h3>
    <table style="border-collapse: collapse !important; width: 100% !important; border: 2px solid blue !important; background-color: white !important;">
        <tr>
            <th style="border: 2px solid blue !important; background-color: #dbe4ff !important; color: black !important; padding: 8px !important;">Колонка 1</th>
            <th style="border: 2px solid blue !important; background-color: #dbe4ff !important; color: black !important; padding: 8px !important;">Колонка 2</th>
        </tr>
        <tr>
            <td style="border: 2px solid blue !important; background-color: white !important; color: black !important; padding: 8px !important;">Ячейка 1</td>
            <td style="border: 2px solid blue !important; background-color: white !important; color: black !important; padding: 8px !important;">Ячейка 2</td>
        </tr>
    </table>
</div>
"""

# ---------------------------------------------------------
# ТЕСТ Б: Классический <style> с уникальным ID (КРАСНАЯ РАМКА)
# Так делает сам Pandas, когда выводит датафреймы
# ---------------------------------------------------------
test_b = """
<div style="margin-bottom: 30px;">
    <h3 style="color: black !important;">Тест Б (Красный): Тег &lt;style&gt; + ID</h3>
    <style>
        #test-b-table { border-collapse: collapse !important; width: 100% !important; border: 2px solid red !important; background-color: white !important; }
        #test-b-table th { border: 2px solid red !important; background-color: #ffdce0 !important; color: black !important; padding: 8px !important; }
        #test-b-table td { border: 2px solid red !important; background-color: white !important; color: black !important; padding: 8px !important; }
    </style>
    <table id="test-b-table">
        <tr><th>Колонка 1</th><th>Колонка 2</th></tr>
        <tr><td>Ячейка 1</td><td>Ячейка 2</td></tr>
    </table>
</div>
"""

# ---------------------------------------------------------
# ТЕСТ В: Shadow DOM (ЗЕЛЕНАЯ РАМКА)
# Прячем HTML от CSS-движка Colab в изолированную "капсулу" через JS
# ---------------------------------------------------------
test_c = """
<div style="margin-bottom: 30px;">
    <h3 style="color: black !important;">Тест В (Зеленый): Изоляция через Shadow DOM</h3>
    <div id="shadow-host-c"></div>
    <script>
        const host = document.getElementById('shadow-host-c');
        if (!host.shadowRoot) {
            const shadow = host.attachShadow({mode: 'open'});
            shadow.innerHTML = `
                <style>
                    table { border-collapse: collapse; width: 100%; border: 2px solid green; background: white; font-family: sans-serif; }
                    th, td { border: 2px solid green; padding: 8px; color: black; }
                    th { background-color: #d1f2eb; }
                </style>
                <table>
                    <tr><th>Колонка 1</th><th>Колонка 2</th></tr>
                    <tr><td>Ячейка 1</td><td>Ячейка 2</td></tr>
                </table>
            `;
        }
    </script>
</div>
"""

# ---------------------------------------------------------
# ТЕСТ Г: Без тега table вообще (ФИОЛЕТОВАЯ РАМКА)
# Если Colab ломает именно таблицы, сверстаем её на div-ах (CSS Grid)
# ---------------------------------------------------------
test_d = """
<div style="margin-bottom: 30px;">
    <h3 style="color: black !important;">Тест Г (Фиолетовый): Вёрстка на DIV-ах (без table)</h3>
    <div style="display: grid; grid-template-columns: 1fr 1fr; border-top: 2px solid purple; border-left: 2px solid purple; background: white; font-family: sans-serif;">
        <!-- Заголовки -->
        <div style="border-right: 2px solid purple; border-bottom: 2px solid purple; background: #e8daef; padding: 8px; color: black; font-weight: bold; text-align: center;">Колонка 1</div>
        <div style="border-right: 2px solid purple; border-bottom: 2px solid purple; background: #e8daef; padding: 8px; color: black; font-weight: bold; text-align: center;">Колонка 2</div>
        <!-- Строка 1 -->
        <div style="border-right: 2px solid purple; border-bottom: 2px solid purple; padding: 8px; color: black; text-align: center;">Ячейка 1</div>
        <div style="border-right: 2px solid purple; border-bottom: 2px solid purple; padding: 8px; color: black; text-align: center;">Ячейка 2</div>
    </div>
</div>
"""

display(HTML(test_a + test_b + test_c + test_d))
```
