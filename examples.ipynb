```python
from IPython.display import display, HTML
import uuid

# Уникальный ID для элемента
elem_id = f"debug-container-{uuid.uuid4()}"

html_code = f"""
<div id="{elem_id}" style="padding: 10px; border: 2px solid red;">
    <h3 style="color: black !important;">Отладка стилей Colab</h3>
    <table id="test-table" style="border: 1px solid black; border-collapse: collapse; width: 100%;">
        <thead>
            <tr>
                <th style="border: 1px solid black; padding: 5px;">Колонка 1</th>
                <th style="border: 1px solid black; padding: 5px;">Колонка 2</th>
            </tr>
        </thead>
        <tbody>
            <tr>
                <td style="border: 1px solid black; padding: 5px; background-color: #f9f9f9;">Тест 1</td>
                <td style="border: 1px solid black; padding: 5px; background-color: #f9f9f9;">Тест 2</td>
            </tr>
            <tr>
                <td style="border: 1px solid black; padding: 5px; background-color: #ffffff;">Тест 3</td>
                <td style="border: 1px solid black; padding: 5px; background-color: #ffffff;">Тест 4</td>
            </tr>
        </tbody>
    </table>
    
    <div id="debug-output" style="margin-top: 15px; background: #eee; padding: 10px; font-family: monospace; white-space: pre-wrap; font-size: 14px; color: black !important;">
        Ожидание данных от браузера...
    </div>
</div>
"""

js_code = f"""
<script>
(function() {{
    setTimeout(function() {{
        const table = document.getElementById('test-table');
        if (!table) return;
        
        const th = table.querySelector('th');
        const td = table.querySelector('td');
        const tbodyRow = table.querySelector('tbody tr');
        
        // Получаем итоговые стили, которые отрисовал браузер (с учетом вмешательства Colab)
        const tableStyle = window.getComputedStyle(table);
        const thStyle = window.getComputedStyle(th);
        const tdStyle = window.getComputedStyle(td);
        const rowStyle = window.getComputedStyle(tbodyRow);

        let debugInfo = "--- АНАЛИЗ СТИЛЕЙ TABLE ---\\n";
        debugInfo += "Background-color: " + tableStyle.backgroundColor + "\\n";
        debugInfo += "Border-collapse: " + tableStyle.borderCollapse + "\\n";
        debugInfo += "Border: " + tableStyle.border + "\\n";
        debugInfo += "Color: " + tableStyle.color + "\\n\\n";

        debugInfo += "--- АНАЛИЗ СТИЛЕЙ TH ---\\n";
        debugInfo += "Background-color: " + thStyle.backgroundColor + "\\n";
        debugInfo += "Border: " + thStyle.border + "\\n";
        debugInfo += "Color: " + thStyle.color + "\\n\\n";

        debugInfo += "--- АНАЛИЗ СТИЛЕЙ TD ---\\n";
        debugInfo += "Background-color: " + tdStyle.backgroundColor + "\\n";
        debugInfo += "Border: " + tdStyle.border + "\\n";
        debugInfo += "Color: " + tdStyle.color + "\\n\\n";
        
        debugInfo += "--- АНАЛИЗ СТИЛЕЙ TR ---\\n";
        debugInfo += "Background-color: " + rowStyle.backgroundColor + "\\n\\n";
        
        // Ищем, какие классы Colab повесил на body
        let bodyClasses = document.body.className;
        debugInfo += "--- АКТИВНЫЕ CSS КЛАССЫ BODY ---\\n";
        debugInfo += (bodyClasses ? bodyClasses : "Нет классов") + "\\n\\n";
        
        // Проверяем наличие специфичных CSS-переменных Colab
        let rootStyle = window.getComputedStyle(document.body);
        debugInfo += "--- НЕКОТОРЫЕ ПЕРЕМЕННЫЕ COLAB ---\\n";
        debugInfo += "--colab-bg-color: " + rootStyle.getPropertyValue('--colab-bg-color') + "\\n";
        debugInfo += "--colab-primary-text-color: " + rootStyle.getPropertyValue('--colab-primary-text-color') + "\\n";
        debugInfo += "--theme-id: " + rootStyle.getPropertyValue('--theme-id') + "\\n";

        document.getElementById('debug-output').textContent = debugInfo;
    }}, 1000); // Ждем 1 секунду, чтобы Colab точно успел применить свои стили
}})();
</script>
"""

display(HTML(html_code + js_code))
```
